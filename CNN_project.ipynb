```python
# 1. Import required libraries
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# 2. Load the Dataset (CIFAR-10 contains RGB images of 10 different categories)
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

# Computers only know pixels 0-255. We divide by 255 to "normalize" them to 0.0-1.0.
# This helps the neural network learn much faster.
train_images, test_images = train_images / 255.0, test_images / 255.0

# 3. Build the CNN (Mapping exactly to the concepts you learned)
model = models.Sequential()

# LAYER 1: Convolution + ReLU (32 filters of 3x3 kernel size)
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)))
# LAYER 2: Max Pooling (Shrinks the image using a 2x2 grid)
model.add(layers.MaxPooling2D((2, 2)))

# LAYER 3 & 4: More Convolution + Pooling to find complex features
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))

# LAYER 5: The Classifier
model.add(layers.Flatten()) # Flattens the 3D grid into a 1D list
model.add(layers.Dense(64, activation='relu')) # A standard neural network layer
model.add(layers.Dense(10)) # 10 output nodes for our 10 categories (dog, plane, etc.)

# 4. Compile the Model (Setting the learning rules)
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# 5. Train the Model (This is where the network actually learns the features)
print("Starting training! This will take a few minutes...")
history = model.fit(train_images, train_labels, epochs=10,
                    validation_data=(test_images, test_labels))

# 6. Plot the Training Results
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label = 'Validation Accuracy')
plt.xlabel('Epoch (Training Rounds)')
plt.ylabel('Accuracy')
plt.legend(loc='lower right')
plt.show()
```
